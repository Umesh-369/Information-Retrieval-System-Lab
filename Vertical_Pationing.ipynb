```python
documents = {
    "Doc1": ["apple", "banana"],
    "Doc2": ["mango", "orange"],
    "Doc3": ["grapes", "pineapple"]
}
```

```python
word_signature = {
    "apple": "100000",
    "banana": "010000",
    "mango": "001000",
    "orange": "000100",
    "grapes": "000010",
    "pineapple": "000001"
}

```

```python
# Bitwise OR
def bitwise_or(sig1, sig2):
    result = ""

    for a, b in zip(sig1, sig2):
        if a == "1" or b == "1":
            result += "1"
        else:
            result += "0"

    return result
```

```python
#Bitwise AND
def bitwise_and(sig1, sig2):
    result = ""

    for a, b in zip(sig1, sig2):
        if a == "1" and b == "1":
            result += "1"
        else:
            result += "0"

    return result

```

```python
document_signatures = {}

print("\nDOCUMENT SIGNATURES\n")

for doc, words in documents.items():
    signature = "000000"

    print(doc)

    for word in words:
        print(word, "->", word_signature[word])
        signature = bitwise_or(signature, word_signature[word])

    document_signatures[doc] = signature

    print("Document Signature:", signature)
    print()
```

```python
print("VERTICAL PARTITIONING\n")

partitions = {}

for doc, sig in document_signatures.items():
    p1 = sig[:3]
    p2 = sig[3:]

    partitions[doc] = (p1, p2)

    print(doc)
    print("Partition 1 :", p1)
    print("Partition 2 :", p2)
    print()
```

```python
query = input("Enter Search Word: ")

query_signature = word_signature[query]

print("\nQuery Signature :", query_signature)

query_p1 = query_signature[:3]
query_p2 = query_signature[3:]

print("Query Partition 1 :", query_p1)
print("Query Partition 2 :", query_p2)
```

```python
if "1" in query_p1:
    partition = 1
    query_part = query_p1
else:
    partition = 2
    query_part = query_p2

print("\nSearching in Partition", partition)

```

```python
print("\nSEARCH RESULT\n")

for doc, (p1, p2) in partitions.items():

    if partition == 1:
        stored = p1
    else:
        stored = p2

    result = bitwise_and(stored, query_part)

    print(doc)
    print("Stored Partition :", stored)
    print("Query Partition  :", query_part)
    print("Bitwise AND      :", result)

    if result == query_part:
        print("Match Found")
        print("Original Document :", " ".join(documents[doc]))
    else:
        print("No Match")

    print()
```
