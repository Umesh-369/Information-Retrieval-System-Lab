```python
documents = {
    "Doc1": ["apple", "banana"],
    "Doc2": ["mango", "orange"],
    "Doc3": ["grapes", "pineapple"]
}

word_signature = {
    "apple": "100000",
    "banana": "010000",
    "mango": "001000",
    "orange": "000100",
    "grapes": "000010",
    "pineapple": "000001"
}
```

```python
# Bitwise OR
def bitwise_or(sig1, sig2):
    result = ""

    for a, b in zip(sig1, sig2):
        if a == "1" or b == "1":
            result += "1"
        else:
            result += "0"

    return result
```

```python
# Compression (Run-Length Encoding)
def compress(signature):
    result = []
    count = 1

    for i in range(1, len(signature)):
        if signature[i] == signature[i - 1]:
            count += 1
        else:
            result.append(f"{signature[i - 1]}×{count}")
            count = 1

    result.append(f"{signature[-1]}×{count}")

    return ",".join(result)

```

```python
# Decompression
def decompress(compressed):
    signature = ""

    parts = compressed.split(",")

    for part in parts:
        bit, count = part.split("×")
        signature += bit * int(count)

    return signature
```

```python
# Bitwise AND
def bitwise_and(sig1, sig2):
    result = ""

    for a, b in zip(sig1, sig2):
        if a == "1" and b == "1":
            result += "1"
        else:
            result += "0"

    return result
```

```python
print("DOCUMENT SIGNATURE CREATION\n")

document_signatures = {}

for doc, words in documents.items():
    doc_sig = "000000"

    print(doc)

    for word in words:
        print(word, "->", word_signature[word])
        doc_sig = bitwise_or(doc_sig, word_signature[word])

    document_signatures[doc] = doc_sig

    print("Document Signature:", doc_sig)
    print()


print("COMPRESSION\n")
```

```python
compressed_file = {}

for doc, sig in document_signatures.items():
    comp = compress(sig)

    compressed_file[doc] = comp

    print(doc)
    print(sig, "->", comp)
    print()
```

```python
query = input("Enter Search Word: ")

query_sig = word_signature[query]

print("\nQuery Signature:", query_sig)

print("\nSEARCH RESULT\n")

for doc, comp in compressed_file.items():
    original = decompress(comp)

    and_result = bitwise_and(original, query_sig)

    print(doc)
    print("Compressed   :", comp)
    print("Decompressed :", original)
    print("Bitwise AND  :", and_result)

    if "1" in and_result:
        print("Match Found")
        print("Original Document:", " ".join(documents[doc]))
    else:
        print("No Match")

    print()
```
